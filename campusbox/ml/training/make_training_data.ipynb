```python
import random
import pandas as pd

random.seed(42)

categories = ["High-Protein", "Asian Rice Bowls", "Vegetarian", "Low-Carb", "Western", "Comfort"]
rows = []

for i in range(800): #800 fake students with random quiz answers
    age = random.randint(17, 30)
    gender = random.choice(["female", "male", "other"])
    is_on_campus = random.choice([True, False])
    budget_per_meal = random.randint(3, 15)
    fitness_goal = random.choice(["lose_weight", "stay_healthy", "build_muscle"])
    spice_tolerance = random.randint(0, 3)
    diet = random.choice(["none", "none", "none", "vegetarian", "vegan", "halal"])

    if diet in ["vegetarian", "vegan"]:
        category = "Vegetarian"
    elif fitness_goal == "build_muscle":
        category = "High-Protein"
    elif fitness_goal == "lose_weight":
        category = "Low-Carb"
    elif budget_per_meal <= 6:
        category = "Comfort"
    elif spice_tolerance >= 2:
        category = "Asian Rice Bowls"
    else:
        category = "Western"
 
    if random.random() < 0.1: #1 in 10 students have a random category to introduce some noise
        category = random.choice(categories)

    rows.append([age, gender, is_on_campus, budget_per_meal, fitness_goal,
                 spice_tolerance, diet, category])

df = pd.DataFrame(rows, columns=["age", "gender", "is_on_campus", "budget_per_meal", #turns list into pandas table and saves to csv file
                                 "fitness_goal", "spice_tolerance", "diet", "category"])
df.to_csv("training_data.csv", index=False)
print(df["category"].value_counts())
df.head()
```
